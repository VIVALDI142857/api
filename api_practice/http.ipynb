```python
base_url = 'http://localhost:8000'
base_url = 'http://127.0.0.1:9123'
```

```python
import requests 
import pandas as pd

url = f'{base_url}/predict_file'

with open("../api_dfs/X_test.csv", "rb") as file:
    files = {"file": ("data.csv", file, "text/csv")}
    response = requests.post(url, files=files)
print(response)
pd.read_json(response.json())

```

```python
url = f'{base_url}/predict_str'

X_test = pd.read_csv('../api_dfs/X_test.csv')
payload = {"data": X_test.to_json(orient='records')}
response = requests.post(url, json=payload)
print(response)
pd.read_json(response.json())
```

```python
url = f'{base_url}/predict_each_col'

X_test = pd.read_csv('../api_dfs/X_test.csv')
payload = {"data": X_test.to_dict(orient='records')}
response = requests.post(url, json=payload)
print(response)
pd.read_json(response.json())
```

```python
y_test = pd.read_csv('y_test.csv')
```

```python
y_pred = pd.read_json(response.json())
```

```python
from sklearn.metrics import fbeta_score
fbeta_score = fbeta_score(y_pred, y_test, beta=1.5)
fbeta_score
```
